```python
import argparse
import logging
import subprocess
from pathlib import Path

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def run_hdfs_command(args):
    """Run an HDFS command and raise a clear error if it fails."""
    try:
        result = subprocess.run(
            ["hdfs", "dfs", *args],
            capture_output=True,
            text=True,
            check=False,
        )
    except FileNotFoundError as exc:
        raise RuntimeError("HDFS CLI not found in PATH. Make sure Hadoop/HDFS is installed and configured.") from exc

    if result.returncode != 0:
        error_message = result.stderr.strip() or result.stdout.strip() or "Unknown HDFS error"
        raise RuntimeError(f"HDFS command failed: {' '.join(['hdfs', 'dfs', *args])}\n{error_message}")

    return result.stdout.strip()


def ensure_hdfs_directory(path):
    """Create a directory in HDFS if it does not exist."""
    run_hdfs_command(["-mkdir", "-p", path])
    logging.info("Ensured HDFS directory exists: %s", path)


def remove_hdfs_path(path):
    """Remove an HDFS path if it exists."""
    try:
        run_hdfs_command(["-rm", "-r", "-f", path])
        logging.info("Removed HDFS path: %s", path)
    except RuntimeError as exc:
        # Some environments may error when the path does not exist.
        # This is usually safe to ignore for cleanup logic.
        logging.warning("Cleanup warning for %s: %s", path, exc)


def load_yellow_taxi(base_path: str, loading_date: str, source_dir: str):
    """Copy the yellow taxi parquet file from local storage into HDFS Bronze."""
    stg_path = f"{base_path}/stg_temp_write/yellow"
    bronze_path = f"{base_path}/bronze/yellow"
    bronze_partition = f"{bronze_path}/date={loading_date}"
    source_file = f"{source_dir}/yellow_tripdata_{loading_date}.parquet"

    # Validate local source file before moving anything
    local_source = Path(source_file)
    if not local_source.exists():
        raise FileNotFoundError(f"Source file not found: {local_source}")

    # Clean old outputs
    remove_hdfs_path(bronze_partition)
    remove_hdfs_path(stg_path)

    # Create required directories
    ensure_hdfs_directory(stg_path)
    ensure_hdfs_directory(bronze_partition)

    # Copy source to staging
    logging.info("Copying local file to staging: %s -> %s", local_source, stg_path)
    run_hdfs_command(["-put", str(local_source), stg_path])

    staged_file = f"{stg_path}/yellow_tripdata_{loading_date}.parquet"
    target_file = f"{bronze_partition}/yellow_tripdata_{loading_date}.parquet"

    # Move from staging to Bronze
    logging.info("Moving file from staging to Bronze: %s -> %s", staged_file, target_file)
    run_hdfs_command(["-mv", staged_file, target_file])

    logging.info("Yellow taxi ingestion completed successfully for date %s", loading_date)


def parse_args():
    repo_root = Path(__file__).resolve().parents[2]
    default_data_dir = str(repo_root / "data")

    parser = argparse.ArgumentParser(description="Ingest Yellow Taxi parquet data into HDFS Bronze.")
    parser.add_argument("--base-path", default="/user/linux/NYC-Taxi", help="Base HDFS path for the data lake.")
    parser.add_argument("--loading-date", default="2019-02", help="Date/month to ingest in YYYY-MM format.")
    parser.add_argument("--source-dir", default=default_data_dir, help="Local source directory containing the parquet file.")
    return parser.parse_args()



def main():
    args = parse_args()
    try:
        load_yellow_taxi(
            base_path=args.base_path,
            loading_date=args.loading_date,
            source_dir=args.source_dir,
        )
    except Exception as exc:
        logging.exception("Yellow taxi ingestion failed: %s", exc)
        raise


if __name__ == "__main__":
    main()
```
